#### The logistic population model:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size
* $x_{\mathrm{next}}$ is next year's population size
* $a$ is the ability to reproduce.



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Logistic Population Model (Basic)
a = 3.4
x = np.zeros(50)
x[0] = 0.5 #Initial population (must be between 0 and 1)
for i in range(1,50): #Because we've already assigned a value to x[0]
  x[i] = a*x[i-1]*(1-x[i-1])
  x[i] = np.maximum(x[i],0) #Replaces a negative population value with zero

plt.plot(x)
plt.xlabel('Generation')
plt.ylabel('Population')
plt.title('Playing God')
plt.xlim([0,50])
plt.ylim([0,1])
```

#### The logistic population model with harvesting:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}}) - h$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size.
* $x_{\mathrm{next}}$ is next year's population size.
* $a$ is the ability to reproduce.
* $h$ is the size of the harvest between 0 and 1.


```python
#Logistic Population Model with Harvesting
a = 3.4
h = 0.18
x = np.zeros(50)
x[0] = 0.5 #Initial population (must be between 0 and 1)
for i in range(1,50): #Because we've already assigned a value to x[0]
  x[i] = a*x[i-1]*(1-x[i-1])-h
  x[i] = np.maximum(x[i],0) #Replaces a negative population value with zero

plt.plot(x)
plt.xlabel('Generation')
plt.ylabel('Population')
plt.title('Playing God')
plt.xlim([0,50])
plt.ylim([0,1])
```

#### The logistic population model with randomness:

$x_{\mathrm{next}} = (a+r) x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size.
* $x_{\mathrm{next}}$ is next year's population size.
* $a$ is the ability to reproduce.
* $r$ is a normally distributed random number that changes each year.

```python
#Logistic Population Model with Randomness
a = 3.8
h = 0.18
x = np.zeros(50)
x[0] = 0.5 #Initial population (must be between 0 and 1)
for i in range(1,50): #Because we've already assigned a value to x[0]
  n = 0.4
  r = np.random.randn()*n #Generate a random number from a normal distribution with a mean of zero and a standard deviation of n
  x[i] = (a+r)*x[i-1]*(1-x[i-1])-h
  x[i] = np.maximum(x[i],0) #Replaces a negative population value with zero

plt.plot(x)
plt.xlabel('Generation')
plt.ylabel('Population')
plt.title('Playing God')
plt.xlim([0,50])
plt.ylim([0,1])
```

#### The logistic map:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size
* $x_{\mathrm{next}}$ is next year's population size
* $a$ is the ability to reproduce.

Follow these steps:

1. Choose a random value of $a$ from 0 to 4.

2. Choose a random starting population x_{\mathrm{initial}} from 0 to 1

3. Run the model for 2,000 generations and record the $x$ value of the last generation.

4. Repeat steps 1-3 a large number of times (thousands)

5. Make a scatter plot of $a$ (x-axis) versus final $x$ (y-axis).



```python
#Logistic Map with Monte Carlo
a = np.random.rand()*4 #Generate a random number between 0 and 4 (reproduction rate)
x = np.random.rand() #Generate a random population size
for i in range(2000): #Compute 2000 times (for 2000 generations)
  x = a*x*(1-x) #This is coding, not maths
```

```python
#With a Lot More Stuff
a_output = [] #Create and record an empty array of a values
x_output = []
for j in range(200000): #Loop within a loop
  a = np.random.rand()*4 #Generate a random number between 0 and 4 (reproduction rate)
  a_output.append(a)
  x = np.random.rand() #Generate a random population size
  for i in range(2000): #Compute 2000 times (for 2000 generations)
    x = a*x*(1-x) #This is coding, not maths
  x_output.append(x) #Let the code run 2000 times, and give the final output
```

```python
plt.plot(a_output, x_output, '.', markersize=0.4)
plt.xlabel('Reproduction Rate')
plt.ylabel('Population after 2000 Generations')
plt.title('Being God')
```

Observations:
- 0 < a < 1 => extinction
- a = 2 => x roughly equal to 0.5
- x > 3 => splits (bifurcation)



```python
#Looking at the interesting part of the graph
a_output = []
x_output = []
for j in range(200000):
  a = np.random.rand()*0.5+3.3 #Restrict to reproduction rates between 3.3 and 3.8
  a_output.append(a)
  x = np.random.rand()
  for i in range(2000):
    x = a*x*(1-x)
  x_output.append(x)

plt.plot(a_output, x_output, '.', markersize=0.5)
plt.xlabel('Reproduction Rate')
plt.ylabel('Population after 2000 Generations')
plt.title('Being God')
```

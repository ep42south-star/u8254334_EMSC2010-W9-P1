#### The logistic population model:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size
* $x_{\mathrm{next}}$ is next year's population size
* $a$ is the ability to reproduce.



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = 3.8
h = 0.0
x = np.zeros(50)
x[0] = 0.5 #Initial population (must be between 0 and 1)
for i in range(1,50): #Because we've already assigned a value to x[0]
  n = 0.4
  r = np.random.randn()*n #Generate a random number from a normal distribution with a mean of zero and a standard deviation of n
  x[i] = (a+r)*x[i-1]*(1-x[i-1])-h
  x[i] = np.maximum(x[i],0) #Replaces a negative population value with zero

plt.plot(x)
plt.xlabel('Generation')
plt.ylabel('Population')
plt.title('Playing God')
plt.xlim([0,50])
plt.ylim([0,1])
```
